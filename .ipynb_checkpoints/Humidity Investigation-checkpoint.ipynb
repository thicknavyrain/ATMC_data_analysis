```python
import pandas as pd
import matplotlib.colors as mcolors
import colorsys
import itertools
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
# outdoor_hum = pd.read_csv('./tagged_data/outdoor_sen_meta_tagged_humidity.csv') 
# outdoor_hum = pd.read_csv('./tagged_data/outdoor_hum_sen_meta_tagged.csv')
outdoor_hum = pd.read_csv('./2023_11_20/outdoor_hum_meta_tagged.csv')

# Convert 'Date/Time' to datetime format
outdoor_hum['Date/Time'] = pd.to_datetime(outdoor_hum['Date/Time'])

# Set 'Date/Time' as the index
outdoor_hum.set_index('Date/Time', inplace=True)
```

```python
site_type = {'AD': ['Asylum Down', 'Commercial/\nbusiness/\nindustrial','blue'],
             'ASH' : ['Ashaiman', 'Medium/low-density\nresidential', 'tab:orange'],
             'EL': ['East Legon', 'Medium/low-density\nresidential', 'tab:orange'],
             'JT' : ['James Town', 'High-density\nresidential', 'red'],
             'LA': ['Labadi', 'Medium/low-density\nresidential', 'tab:orange'],
             'N1W': ['N1 West Motorway', 'Commercial/\nbusiness/\nindustrial','blue'],
             'NM': ['Nima', 'High-density\nresidential', 'red'],
             'TF': ['Taifa','Medium/low-density\nresidential', 'tab:orange'],
             'TMW': ['Tema Motorway', 'Commercial/\nbusiness/\nindustrial','blue'],
             'UGH': ['University of Ghana', 'Peri-urban background', 'green']}
```

```python
# Convert 'Date/Time' to datetime format
# outdoor_hum['Date/Time'] = pd.to_datetime(outdoor_hum['Date/Time'])

# Set 'Date/Time' as the index
# outdoor_hum.set_index('Date/Time', inplace=True)

# Resample the data to hourly, taking the mean humidity measurement within each hour
# If you want to use different aggregate function, you can replace .mean() with .sum(), .min(), .max(), etc.
outdoor_hum_hourly = outdoor_hum.groupby('site_id').resample('H').mean(numeric_only=True)

# Reset the index
outdoor_hum_hourly.reset_index(inplace=True)

# Pivot the data so that each sensor has its own column
outdoor_hum_pivot = outdoor_hum_hourly.pivot(index='Date/Time', columns='site_id', values='Value')

# Compute the correlation between each pair of sensors
correlations = outdoor_hum_pivot.corr()

print(correlations)
```

```python
# Function to convert colormap
def convert_blue_to_green(cmap):
    cmap = plt.get_cmap(cmap)
    colors_rgb = cmap(np.arange(cmap.N))

    # Convert RGB to HLS
    colors_hls = np.array([colorsys.rgb_to_hls(*color_rgb) for color_rgb in colors_rgb[:, :3]])

    # Change hue from blue to green (240 degrees to 120 degrees)
    colors_hls[:, 0] = (colors_hls[:, 0] - 2 / 3) % 1

    # Convert back to RGB
    colors_rgb = np.array([colorsys.hls_to_rgb(*color_hls) for color_hls in colors_hls])

    # Create new colormap
    new_cmap = mcolors.LinearSegmentedColormap.from_list("green_seismic", colors_rgb, cmap.N)

    return new_cmap
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Plot size
fig, ax = plt.subplots(figsize=(10,10))

# Assume the sensor names are 'Sensor 1', 'Sensor 2', etc.
site_names = correlations.columns
labelling = {name: name for name in site_names}

# Your function to create a new colormap
new_cmap = convert_blue_to_green('seismic')

# Create a mask to hide the upper triangle of the correlation matrix, since it's symmetric
mask = np.triu(np.ones_like(correlations, dtype=bool))

for i in range(len(mask)):
    mask[i,i] = False

# Format the correlation values to 3 decimal places for display
labels = (np.asarray(["{:.3f}".format(data) for data in correlations.to_numpy().flatten()])).reshape(len(site_names),len(site_names))

# Rename the columns and index of the correlation matrix for display
correlations.rename(columns=labelling, index=labelling, inplace=True)

# Plot the correlation matrix
corr_plot = sns.heatmap(correlations.astype(float), annot=labels, fmt='', cmap=plt.get_cmap(new_cmap), vmin=-1, vmax=1, cbar=False, ax=ax, mask=mask)

# Fix for the matplotlib bug that cuts off half of the top and bottom cells
b, t = plt.ylim()
t -= 0.5
plt.ylim(b, t)

plt.xticks(rotation=90)
plt.tight_layout()
plt.xlabel('Site')
plt.ylabel('Site')
plt.title('Pearson corrleation of relative humidity measurements between sites at a given hour')

# Save the plot
plt.savefig('humidity_correlation_plot.png', facecolor='white')

# Display the plot
plt.show()
```

```python
# Create a new dataframe where each site's hourly readings are separate columns
outdoor_hum_hourly = outdoor_hum.groupby('site_id').resample('H').mean(numeric_only=True).unstack(level=0)['Value']

# Initialize a new dataframe to store the differences
diffs = pd.DataFrame(index=outdoor_hum_hourly.columns, columns=outdoor_hum_hourly.columns)

# Calculate the average absolute difference for each pair of sites
for site1, site2 in itertools.combinations(outdoor_hum_hourly.columns, 2):
    diffs.loc[site1, site2] = diffs.loc[site2, site1] = np.mean(np.abs(outdoor_hum_hourly[site1] - outdoor_hum_hourly[site2]))

# Convert the differences dataframe to float
diffs = diffs.astype(float)

# Assume the site names are 'site 1', 'site 2', etc.
site_names = diffs.columns

labelling = {name: name for name in site_names}

# Rename the columns and index of the differences dataframe for display
diffs.rename(columns=labelling, index=labelling, inplace=True)

# Create a mask to hide the upper triangle of the correlation matrix, since it's symmetric
mask = np.triu(np.ones_like(diffs, dtype=bool))

for i in range(len(mask)):
    mask[i,i] = False

# Plot size
fig, ax = plt.subplots(figsize=(10,10))

# Format the difference values to 3 decimal places for display
labels = (np.asarray(["{:.3f}".format(data) for data in diffs.to_numpy().flatten()])).reshape(len(site_names),len(site_names))

# Your function to create a new colormap
# new_cmap = convert_blue_to_green('seismic')

# Plot the difference matrix
# diff_plot = sns.heatmap(diffs.astype(float), annot=labels, fmt='', cmap=plt.get_cmap(new_cmap), vmin=0, vmax=100, cbar=False, ax=ax, mask=mask)
diff_plot = sns.heatmap(diffs.astype(float), annot=labels, fmt='', cmap='seismic', cbar=False, ax=ax, mask=mask)

# Fix for the matplotlib bug that cuts off half of the top and bottom cells
b, t = plt.ylim()
t -= 0.5
plt.ylim(b, t)

plt.title('Average difference in relative humidity measurements (absolute %) between sites at a given hour')

plt.xticks(rotation=90)

# Get the original labels
ylabels = [item.get_text() for item in ax.get_yticklabels()]
xlabels = [item.get_text() for item in ax.get_xticklabels()]

# Replace the first ylabel and the last xlabel with an empty string
ylabels[0] = ''
xlabels[-1] = ''

# Set the new labels
ax.set_yticklabels(ylabels)
ax.set_xticklabels(xlabels, rotation=90)

plt.tight_layout()
plt.savefig('humidity_difference_plot.png', facecolor='white')
```

```python
# Reset index
# outdoor_hum.reset_index(inplace=True)
# indoor_hum.reset_index(inplace=True)

# Sort both dataframes by 'Date/Time'
outdoor_hum.sort_values('Date/Time', inplace=True)
indoor_hum.sort_values('Date/Time', inplace=True)

# Rename 'Value' column in indoor dataframe to avoid confusion after merge
indoor_hum.rename(columns={'Value': 'Indoor_Value', 'redcap_repeat_instance_y' : 'redcap_y', 'record_id_y': 'record_y', 'redcap_repeat_instrument_y': 'redcap_instrument_y'}, inplace=True)

# Merge the dataframes
merged_df = pd.merge_asof(outdoor_hum, indoor_hum, 
                          on='Date/Time', by='site_id', 
                          tolerance=pd.Timedelta('1 hour'), direction='nearest', suffixes=('_outdoor', '_indoor'))

# Set 'Date/Time' back as index
merged_df.set_index('Date/Time', inplace=True)

common_sites = set(indoor_hum['site_id']).intersection(set(outdoor_hum['site_id']))

for site_id in common_sites:
    data = merged_df[merged_df['site_id'] == site_id]
    
    # Drop NaN values only in 'Value' and 'Indoor_Value'
    data = data[['Value', 'Indoor_Value']].dropna()
    
    print(f"Processing site {site_id}, {len(data)} data points available")
    
    if len(data) < 2:
        print(f"Skipping site {site_id} due to insufficient data")
        continue

    # Determine the color for each row based on the month
    def get_color(date):
        if 11 <= date.month <= 12 or date.month == 1:
            return 'darkred'
        else:
            return 'blue'

    colors = data.index.map(get_color)

    plt.figure(figsize=(6, 6)) # make the plot square

    # Calculate line of best fit
    slope, intercept, r_value, _, _ = stats.linregress(data['Value'], data['Indoor_Value'])
    
    # Generate x values and corresponding y values for the line of best fit
    x = np.linspace(data['Value'].min(), data['Value'].max(), 100)
    y = slope * x + intercept
    
    # Plot the line of best fit without a label
    plt.plot(x, y, color='red')
    
    # Plot the x=y line
    plt.plot([data['Value'].min(), data['Value'].max()], [data['Value'].min(), data['Value'].max()], color='blue', linestyle='--')
    
    # Scatter plot of the data
    # plt.scatter(data['Value'], data['Indoor_Value'], color='green', alpha=0.1)
     # Scatter plot of the data with colors based on Harmattan period
    plt.scatter(data['Value'], data['Indoor_Value'], c=colors, alpha=0.1)   

    plt.title(f'Site {site_id} Indoor vs Outdoor Humidity')
    plt.xlabel('Outdoor Relative Humidity (%)')
    plt.ylabel('Indoor Relative Humidity (%)')
    
    correlation, _ = stats.pearsonr(data['Value'], data['Indoor_Value'])
    print(f'Pearson correlation for site {site_id}: {correlation}')
    
    # Add label with line equation and correlation
    plt.text(0.05, 0.95, f'Pearson r = {correlation:.2f}\ny = {slope:.2f}x + {intercept:.2f}', 
             transform=plt.gca().transAxes, fontsize=12, verticalalignment='top')
    
#     # Find the global maximum and minimum values
#     global_min = min(data['Value'].min(), data['Indoor_Value'].min())
#     global_max = max(data['Value'].max(), data['Indoor_Value'].max())
    
    # Set the limits for both axes
    plt.xlim(50, 100)
    plt.ylim(50, 100)
    
    plt.gca().set_aspect('equal', adjustable='box') # Ensure equal scaling for x and y axes
    plt.grid(True)
#     plt.legend()
    plt.savefig('humidity_indoor_outdoor_plot_'+site_id+'.png', facecolor='white')

```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy import stats

# Assuming merged_df is already created with the necessary columns

for site_id in common_sites:
    site_data = merged_df[merged_df['site_id'] == site_id]

    # Iterate over each indoor sensor at this site
    for indoor_sensor in site_data['sensor_id_indoor'].unique():
        data = site_data[site_data['sensor_id_indoor'] == indoor_sensor]

        # Drop NaN values only in 'Value' and 'Indoor_Value'
        data = data[['Value', 'Indoor_Value']].dropna()
        
        print(f"Processing site {site_id}, sensor {indoor_sensor}, {len(data)} data points available")

        if len(data) < 2:
            print(f"Skipping site {site_id}, sensor {indoor_sensor} due to insufficient data")
            continue

        colors = data.index.map(get_color)

        plt.figure(figsize=(6, 6))  # make the plot square

        # Calculate line of best fit
        slope, intercept, r_value, _, _ = stats.linregress(data['Value'], data['Indoor_Value'])

        # Generate x values and corresponding y values for the line of best fit
        x = np.linspace(data['Value'].min(), data['Value'].max(), 100)
        y = slope * x + intercept

        # Plot the line of best fit without a label
        plt.plot(x, y, color='red')

        # Plot the x=y line
        plt.plot([data['Value'].min(), data['Value'].max()], [data['Value'].min(), data['Value'].max()], color='blue', linestyle='--')

        # Scatter plot of the data with colors based on Harmattan period
        plt.scatter(data['Value'], data['Indoor_Value'], c=colors, alpha=0.1)   

        plt.title(f'Site {site_id} Sensor {indoor_sensor} Indoor vs Outdoor Humidity')
        plt.xlabel('Outdoor Relative Humidity (%)')
        plt.ylabel('Indoor Relative Humidity (%)')

        correlation, _ = stats.pearsonr(data['Value'], data['Indoor_Value'])
        print(f'Pearson correlation for site {site_id}, sensor {indoor_sensor}: {correlation}')

        # Add label with line equation and correlation
        plt.text(0.05, 0.95, f'Pearson r = {correlation:.2f}\ny = {slope:.2f}x + {intercept:.2f}', 
                 transform=plt.gca().transAxes, fontsize=12, verticalalignment='top')

        # Set the limits for both axes
        plt.xlim(50, 100)
        plt.ylim(50, 100)

        plt.gca().set_aspect('equal', adjustable='box')  # Ensure equal scaling for x and y axes
        plt.grid(True)
        plt.savefig(f'humidity_plots/humidity_indoor_outdoor_plot_{site_id}_{indoor_sensor}.png', facecolor='white')
        # plt.close()  # Close the figure to free memory
```

```python
merged_df_no_nan =  merged_df[['Value', 'Indoor_Value']].dropna()
```

```python
from sklearn.metrics import mean_squared_error
from math import sqrt
import pandas as pd
from scipy import stats

sensor_results = []
sensor_order = []

# Assuming merged_df is already created with the necessary columns

for site_id in common_sites:
    site_data = merged_df[merged_df['site_id'] == site_id]

    # Iterate over each indoor sensor at this site
    for indoor_sensor in site_data['sensor_id_indoor'].unique():
        data = site_data[site_data['sensor_id_indoor'] == indoor_sensor]

        # Drop NaN values only in 'Value' and 'Indoor_Value'
        data = data[['Value', 'Indoor_Value']].dropna()
        if(len(data) > 0):
            print(f"Processing site {site_id}, sensor {indoor_sensor}, {len(data)} data points available")
    
            if len(data) < 2:
                print(f"Skipping site {site_id}, sensor {indoor_sensor} due to insufficient data")
                continue
    
            # Calculate line of best fit
            slope, intercept, r_value, p_value, std_err = stats.linregress(data['Value'], data['Indoor_Value'])
    
            # Calculate the predicted values
            predicted_values = slope * data['Value'] + intercept
    
            # Calculate RMSE
            rmse = sqrt(mean_squared_error(data['Indoor_Value'], predicted_values))
    
            sensor_results.append((site_id, indoor_sensor, slope, intercept, r_value, rmse, std_err))
            sensor_order.append(f"{site_id}_{indoor_sensor}")

sensor_results_df = pd.DataFrame(sensor_results, columns=['site_id', 'indoor_sensor_id', 'slope', 'intercept', 'r_value', 'rmse', 'std_err'], index=sensor_order)

# Now let's calculate the "universal" slope and intercept
universal_slope = sensor_results_df['slope'].mean()
universal_intercept = sensor_results_df['intercept'].mean()
print('Universal slope:', universal_slope)
print('Universal intercept:', universal_intercept)

# Now calculate RMSE for all data using the "universal" equation
all_predicted_values = universal_slope * merged_df_no_nan['Value'] + universal_intercept
universal_rmse = sqrt(mean_squared_error(merged_df_no_nan['Indoor_Value'], all_predicted_values))
print('Universal RMSE:', universal_rmse)

# And now for the global fit
global_slope, global_intercept, _, _, _ = stats.linregress(merged_df_no_nan['Value'], merged_df_no_nan['Indoor_Value'])
print('Global slope:', global_slope)
print('Global intercept:', global_intercept)

# Calculate RMSE for all data using the "global" equation
all_predicted_values = global_slope * merged_df_no_nan['Value'] + global_intercept
global_rmse = sqrt(mean_squared_error(merged_df_no_nan['Indoor_Value'], all_predicted_values))
print('Global RMSE:', global_rmse)
```

```python
# First, resample the original data to half-hour intervals using the mean.
# This assumes that your original 'outdoor_hum' dataframe has a datetime index
half_hourly = outdoor_hum.groupby('site_id').resample('30T').mean(numeric_only=True)

# Now, unstack the 'site_id' level of the MultiIndex so we have one column per site
half_hourly = half_hourly.unstack(level=0)['Value']

# Calculate the mean and standard deviation across all sites for each half-hour
citywide_humidity = half_hourly.mean(axis=1)
citywide_std = half_hourly.std(axis=1)

# Combine these into a single dataframe
citywide_humidity_df = pd.DataFrame({
    'Average Humidity': citywide_humidity,
    'Standard Deviation': citywide_std
})
```
